## 1. Problem Overview

The goal of this project is to predict daily sales for Rossmann stores using historical store-level sales records and additional store metadata. This is a supervised machine learning regression problem because the target variable, Sales, is a continuous numerical value, and the model is trained using past observations where the true sales values are already known.

Accurate sales forecasting is an important business task in retail. Reliable predictions can help companies improve inventory management, staff scheduling, promotion planning, and financial decision-making. If future sales are underestimated, stores may run out of stock and lose revenue. If they are overestimated, stores may overstock products and increase operational costs. For this reason, building a strong sales forecasting model has direct practical value.

The Rossmann dataset combines two important sources of information. The first contains daily operational data for each store, including variables such as the date, whether the store was open, whether a promotion was active, the day of the week, holidays, and the number of customers. The second contains store-level attributes, such as store type, assortment type, competition distance, and long-term promotion program information. By combining these two datasets, we can better understand both the short-term and structural factors that influence sales.

This problem is challenging because sales depend on many interacting factors. They vary across time due to seasonality, weekdays, weekends, holidays, and promotional events, and they also vary across stores because different stores have different formats, assortments, and competitive environments. In addition, some variables contain missing values, which makes data preprocessing and feature engineering essential parts of the project.

The overall objective of this notebook is to build an end-to-end machine learning pipeline that can learn the relationship between these variables and store sales, compare different regression models, evaluate their performance using suitable error metrics, and finally generate predictions for unseen test data. In this way, the project demonstrates how machine learning can be used to solve a real-world retail forecasting problem.

## 5. Data Preprocessing

```python
def preprocess_rossmann(data: pd.DataFrame, competition_distance_fill: float) -> pd.DataFrame:
    """Create date features and fill stable missing values."""
    data = data.copy()

    data["Date"] = pd.to_datetime(data["Date"])
    data["Day"] = data["Date"].dt.day
    data["Month"] = data["Date"].dt.month
    data["Year"] = data["Date"].dt.year
    data["WeekOfYear"] = data["Date"].dt.isocalendar().week.astype(int)
    data["DayOfYear"] = data["Date"].dt.dayofyear
    data["IsWeekend"] = data["DayOfWeek"].isin([6, 7]).astype(int)
    data["IsMonthStart"] = data["Date"].dt.is_month_start.astype(int)
    data["IsMonthEnd"] = data["Date"].dt.is_month_end.astype(int)

    data["StateHoliday"] = data["StateHoliday"].fillna("0").astype(str)
    data["IsStateHoliday"] = (data["StateHoliday"] != "0").astype(int)

    data["PromoInterval"] = data["PromoInterval"].fillna("None")
    data["StoreType"] = data["StoreType"].fillna("unknown")
    data["Assortment"] = data["Assortment"].fillna("unknown")

    data["Open"] = data["Open"].fillna(1)
    data["CompetitionDistance"] = data["CompetitionDistance"].fillna(competition_distance_fill)

    for col in [
        "CompetitionOpenSinceMonth",
        "CompetitionOpenSinceYear",
        "Promo2SinceWeek",
        "Promo2SinceYear",
    ]:
        data[col] = data[col].fillna(0)

    return data
```

This function is used to prepare the Rossmann data before giving it to a machine learning model.

I first make a copy of the dataset so that the original data stays unchanged. Then I convert the `Date` column into a proper date format. This allows me to create new useful columns such as the day, month, year, week of the year, and day of the year.

I also create extra features that may help the model understand sales patterns better, such as whether the day is a weekend, whether it is the start of the month, or whether it is the end of the month. These details can be important because customer behavior can change depending on the day or period of the month.

After that, I handle missing values. For example, if there is no state holiday, I replace missing values with `"0"` and create a new column called `IsStateHoliday` to show whether the day is a holiday or not.

For columns like `PromoInterval`, `StoreType`, and `Assortment`, I replace missing values with simple labels such as `"None"` or `"unknown"`. This keeps the information clear and avoids problems later when training the model.

For `CompetitionDistance`, I fill missing values using a value passed into the function, usually the median distance from the training data.

Finally, some competition and promotion date columns have missing values, so I replace them with `0` to show that the information is not available.

## Analysis
Here I create the training and validation datasets.

First, I make a copy of the full dataset so I can work on it without changing the original one.

Then I find the latest date in the dataset. Since this is sales data over time, I do not split the data randomly. Instead, I use the most recent 12 weeks as the validation set, and all the older data as the training set.

This is more realistic because in real life the model is trained on past data and used to predict future sales.

The code then separates the data into:

- `train_df`: all rows before the last 12 weeks
- `valid_df`: the last 12 weeks of data

After that, I calculate the median value of `CompetitionDistance` using only the training data. This value is later used to fill missing competition distances in both the training and validation sets.

I calculate it from the training data only to avoid data leakage. This means I do not allow information from the validation period to influence the training process.

If the median value is missing for some reason, I replace it with `0.0` as a safe fallback.

Finally, I apply the preprocessing function to both datasets using the same competition distance fill value. At the end, I print the shape and date range of both sets to check that the split was done correctly.

## 7. Feature Engineering

```python
MEAN_FEATURE_SPECS = [
    (["Store"], "MeanSales_ByStore"),
    (["Store", "DayOfWeek"], "MeanSales_ByStore_DayOfWeek"),
    (["Store", "Month"], "MeanSales_ByStore_Month"),
    (["Store", "Promo"], "MeanSales_ByStore_Promo"),
    (["Store", "SchoolHoliday"], "MeanSales_ByStore_SchoolHoliday"),
    (["Store", "StateHoliday"], "MeanSales_ByStore_StateHoliday"),
    (["Store", "DayOfWeek", "Promo"], "MeanSales_ByStore_DayOfWeek_Promo"),
    (["StoreType"], "MeanSales_ByStoreType"),
    (["StoreType", "DayOfWeek"], "MeanSales_ByStoreType_DayOfWeek"),
    (["Assortment"], "MeanSales_ByAssortment"),
    (["Promo"], "MeanSales_ByPromo"),
    (["DayOfWeek"], "MeanSales_ByDayOfWeek"),
    (["Month"], "MeanSales_ByMonth"),
]

EXPECTED_MEAN_COLS = [new_col for _, new_col in MEAN_FEATURE_SPECS]

def add_group_mean_from_source(source_df, target_df, group_cols, new_col):
    source_df = source_df.copy()
    target_df = target_df.copy()

    missing_cols = [c for c in group_cols if c not in source_df.columns or c not in target_df.columns]
    if missing_cols:
        target_df[new_col] = np.nan
        return target_df

    stats = (
        source_df.groupby(group_cols, dropna=False)["Sales"]
        .mean()
        .reset_index()
        .rename(columns={"Sales": new_col})
    )
    target_df = target_df.merge(stats, on=group_cols, how="left")
    return target_df

def apply_backoff_fills(df_part, global_mean):
    df_part = df_part.copy()

    for col in EXPECTED_MEAN_COLS:
        if col not in df_part.columns:
            df_part[col] = np.nan

    df_part["MeanSales_ByStore"] = df_part["MeanSales_ByStore"].fillna(global_mean)
    df_part["MeanSales_ByStore_DayOfWeek"] = df_part["MeanSales_ByStore_DayOfWeek"].fillna(df_part["MeanSales_ByStore"])
    df_part["MeanSales_ByStore_Month"] = df_part["MeanSales_ByStore_Month"].fillna(df_part["MeanSales_ByStore"])
    df_part["MeanSales_ByStore_Promo"] = df_part["MeanSales_ByStore_Promo"].fillna(df_part["MeanSales_ByStore"])
    df_part["MeanSales_ByStore_SchoolHoliday"] = df_part["MeanSales_ByStore_SchoolHoliday"].fillna(df_part["MeanSales_ByStore"])
    df_part["MeanSales_ByStore_StateHoliday"] = df_part["MeanSales_ByStore_StateHoliday"].fillna(df_part["MeanSales_ByStore"])
    df_part["MeanSales_ByStore_DayOfWeek_Promo"] = df_part["MeanSales_ByStore_DayOfWeek_Promo"].fillna(
        df_part["MeanSales_ByStore_DayOfWeek"]
    )

    df_part["MeanSales_ByStoreType"] = df_part["MeanSales_ByStoreType"].fillna(df_part["MeanSales_ByStore"])
    df_part["MeanSales_ByStoreType_DayOfWeek"] = df_part["MeanSales_ByStoreType_DayOfWeek"].fillna(
        df_part["MeanSales_ByStoreType"]
    )
    df_part["MeanSales_ByAssortment"] = df_part["MeanSales_ByAssortment"].fillna(df_part["MeanSales_ByStore"])
    df_part["MeanSales_ByPromo"] = df_part["MeanSales_ByPromo"].fillna(df_part["MeanSales_ByStore"])
    df_part["MeanSales_ByDayOfWeek"] = df_part["MeanSales_ByDayOfWeek"].fillna(df_part["MeanSales_ByStore"])
    df_part["MeanSales_ByMonth"] = df_part["MeanSales_ByMonth"].fillna(df_part["MeanSales_ByStore"])

    return df_part

def add_train_only_mean_features(train_part, apply_part):
    train_part = train_part.copy()
    apply_part = apply_part.copy()

    global_mean = train_part["Sales"].mean()

    for group_cols, new_col in MEAN_FEATURE_SPECS:
        apply_part = add_group_mean_from_source(train_part, apply_part, group_cols, new_col)

    apply_part = apply_backoff_fills(apply_part, global_mean)
    return apply_part
```

### Creating train-only mean sales features

In this step, I create extra features based on the average sales of different groups.

The idea is that past sales behavior can give the model useful information. For example, some stores usually have higher sales than others, some days of the week are stronger, and promotions or holidays can change sales patterns.

The `MEAN_FEATURE_SPECS` list defines which grouped averages should be created. For example, the code calculates average sales by:

- store
- store and day of week
- store and month
- store and promotion status
- store and school holiday
- store and state holiday
- store type
- assortment type
- day of week
- month

Each grouped average becomes a new feature, such as `MeanSales_ByStore` or `MeanSales_ByStore_DayOfWeek`.

The function `add_group_mean_from_source` calculates these averages from a source dataset and merges them into another dataset. In this project, the source should be the training data. This is important because the validation data should not be used to create these averages.

The function `apply_backoff_fills` handles cases where a grouped average is missing. For example, if there is no exact average for a specific store, day, and promo combination, the code falls back to a more general average such as the store average. If that is also missing, it uses the global average sales from the training data.

Finally, `add_train_only_mean_features` brings everything together. It calculates the global mean sales from the training data, creates all grouped mean features using only the training data, applies the fallback filling strategy, and returns the dataset with the new features.

Overall, these features help the model learn historical sales patterns while avoiding data leakage from the validation set.

```python
def add_oof_mean_features_time_based(train_part, n_folds=5, min_train_dates=1):
    train_part = train_part.sort_values("Date").copy()
    train_part["_row_id_tmp"] = np.arange(len(train_part))

    unique_dates = np.array(sorted(train_part["Date"].unique()))
    n_dates = len(unique_dates)

    if n_dates < 2:
        for col in EXPECTED_MEAN_COLS:
            train_part[col] = np.nan
        return train_part.drop(columns=["_row_id_tmp"])

    n_folds = min(n_folds, n_dates - 1)
    n_folds = max(n_folds, 2)

    fold_boundaries = np.linspace(0, n_dates, n_folds + 1, dtype=int)
    out_parts = []

    for fold_idx in range(1, n_folds + 1):
        valid_start_idx = fold_boundaries[fold_idx - 1]
        valid_end_idx = fold_boundaries[fold_idx]

        valid_dates = unique_dates[valid_start_idx:valid_end_idx]
        if len(valid_dates) == 0:
            continue

        train_dates_for_fold = unique_dates[:valid_start_idx]
        if len(train_dates_for_fold) < min_train_dates:
            continue

        fold_train = train_part[train_part["Date"].isin(train_dates_for_fold)].copy()
        fold_valid = train_part[train_part["Date"].isin(valid_dates)].copy()

        fold_valid = add_train_only_mean_features(fold_train, fold_valid)
        out_parts.append(fold_valid)

    covered_ids = set()
    if out_parts:
        covered_ids = set(pd.concat(out_parts, axis=0)["_row_id_tmp"].tolist())

    missing_part = train_part[~train_part["_row_id_tmp"].isin(covered_ids)].copy()

    if len(missing_part) > 0:
       
        earliest_missing_date = missing_part["Date"].min()
        hist = train_part[train_part["Date"] < earliest_missing_date].copy()

        if len(hist) == 0:
            for col in EXPECTED_MEAN_COLS:
                missing_part[col] = np.nan
        else:
            missing_part = add_train_only_mean_features(hist, missing_part)

        out_parts.append(missing_part)

    if not out_parts:
        for col in EXPECTED_MEAN_COLS:
            train_part[col] = np.nan
        return train_part.drop(columns=["_row_id_tmp"])

    train_oof = pd.concat(out_parts, axis=0).sort_values("_row_id_tmp").copy()
    train_oof = train_oof.drop(columns=["_row_id_tmp"])

    return train_oof


train_df = add_oof_mean_features_time_based(train_df, n_folds=5)
valid_df = add_train_only_mean_features(train_df, valid_df)

print("OOF train mean features added:", len([c for c in EXPECTED_MEAN_COLS if c in train_df.columns]))
print("Valid mean features added:", len([c for c in EXPECTED_MEAN_COLS if c in valid_df.columns]))
```

### Creating leakage-safe mean features for training and validation

In this step, I create mean sales features in a way that avoids data leakage.

The previous mean features are based on historical average sales, such as average sales by store, weekday, month, promotion status, and other groups. These features are useful, but they must be created carefully. If the model is trained using averages calculated from the same rows it is trying to predict, then the target value `Sales` leaks into the features.

To avoid this problem, I use a time-based out-of-fold method for the training data.

The function `add_oof_mean_features_time_based` first sorts the training data by date. Then it splits the available dates into several time-based folds. For each fold, it uses only earlier dates to calculate the mean sales features, and then applies those features to the next validation-like time block.

This means each training row receives mean features calculated only from past data, not from its own sales value or future sales values.

Very early rows may not have enough previous history. In that case, the code keeps the mean features as missing values or fills them only if earlier historical data exists. This keeps the process honest and prevents leakage.

After creating these out-of-fold mean features for the training set, the validation set is handled differently. Since validation represents future unseen data, its mean features are calculated using the full training data only. This is realistic because, in a real prediction setting, we would already know the historical training data but not the future validation sales.

Finally, the print statements check how many expected mean features were successfully added to the training and validation datasets.

Overall, this step creates powerful historical sales features while making sure the model does not accidentally learn from future information.

## 8. Model Evaluation Setup


```python
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


tscv = TimeSeriesSplit(n_splits=3)

def evaluate_model(y_true, y_pred, model_name="Model"):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    
    wape = (np.abs(y_true - y_pred).sum() / np.abs(y_true).sum()) * 100
    
    
    mask = y_true != 0
    rmspe = np.sqrt(np.mean(((y_true[mask] - y_pred[mask]) / y_true[mask]) ** 2)) * 100
    
    print(model_name)
    print("MAE  :", round(mae, 3))
    print("RMSE :", round(rmse, 3))
    print("R2   :", round(r2, 4))
    print("WAPE :", round(wape, 2), "%")
    print("RMSPE:", round(rmspe, 2), "%")
    
    return {
        "Model": model_name,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2,
        "WAPE": wape,
        "RMSPE": rmspe
    }
```

### Model evaluation setup

In this step, I set up the tools needed to evaluate the performance of the models.

I use `TimeSeriesSplit` instead of a random split because this project is based on sales data over time. This keeps the evaluation closer to a real forecasting situation, where the model is trained on earlier data and tested on later data.

The `evaluate_model` function takes the true sales values, the predicted sales values, and the model name. It then calculates several performance metrics:

- **MAE** measures the average absolute error between the real and predicted sales values.
- **RMSE** gives more weight to larger errors, so it is useful when big mistakes are especially important.
- **R²** shows how much of the variation in sales is explained by the model.
- **WAPE** shows the total prediction error as a percentage of total actual sales.
- **RMSPE** measures the percentage prediction error and is especially relevant for the Rossmann sales problem.


At the end, the function prints the results in a clear format and also returns them as a dictionary. This makes it easier to compare different models later.

```python
target_col = "Sales"
drop_cols = ["Sales", "Date", "Customers"]

feature_cols = [c for c in train_df.columns if c not in ["Sales", "Date", "Customers"]]
feature_cols = [c for c in feature_cols if c in valid_df.columns]

X_train = train_df[feature_cols].copy()
y_train = train_df["Sales"].copy()

X_val = valid_df[feature_cols].copy()
y_val = valid_df["Sales"].copy()


categorical_cols = ['StateHoliday', 'StoreType', 'Assortment', 'PromoInterval']

preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)
    ],
    remainder='passthrough'
)

print(f"Number of features: {len(feature_cols)}")
```

### Preparing features for modelling

In this step, I separate the data into input features and the target variable.

The target column is `Sales`, because this is the value I want the model to predict.

I remove `Sales`, `Date`, and `Customers` from the feature list. `Sales` is removed because it is the target variable. `Date` is removed because I already created useful date-based features earlier, such as month, day, and week of year. `Customers` is also removed because it would not usually be known in advance when predicting future sales, so using it could make the model unrealistically strong.

After that, I make sure that the selected features exist in both the training and validation datasets. This helps avoid errors when fitting and evaluating the model.

Then I create:

- `X_train`: training features
- `y_train`: training target values
- `X_val`: validation features
- `y_val`: validation target values

Next, I define the categorical columns: `StateHoliday`, `StoreType`, `Assortment`, and `PromoInterval`. These columns contain text categories, so they need to be converted into numerical format before being used by most machine learning models.

I use `OneHotEncoder` for this. It creates separate binary columns for each category. The option `handle_unknown='ignore'` makes sure that the model does not fail if a new category appears in the validation data.

The `ColumnTransformer` applies one-hot encoding only to the categorical columns and keeps all remaining numerical columns unchanged.

Finally, I print the number of selected features to check how many input variables will be used by the model.

## 9. Baseline Model

```python
store_avg_sales = train_df.groupby("Store")["Sales"].mean()

baseline_preds = valid_df["Store"].map(store_avg_sales)


baseline_preds = baseline_preds.fillna(train_df["Sales"].mean())

baseline_results = evaluate_model(
    y_true=valid_df["Sales"],
    y_pred=baseline_preds,
    model_name="Baseline"
)
```

### Baseline model

In this step, I create a simple baseline model.

The baseline predicts sales using the average sales of each store from the training data. For example, if Store 10 usually had a certain average sales value in the past, the baseline uses that average as the prediction for Store 10 in the validation set.

I calculate these store averages using the training data only. This is important because the validation data represents future unseen data, so its sales values should not be used when creating the predictions.

If a store appears in the validation set but does not exist in the training set, the code uses the overall average sales from the training data as a fallback.

Finally, I evaluate the baseline predictions using the same `evaluate_model` function. This gives me a reference point to compare more advanced machine learning models against.

Overall, this baseline helps answer an important question: do the more complex models actually perform better than a simple store-average prediction?

## 10. Decision Tree Model

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.pipeline import Pipeline

dt_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', DecisionTreeRegressor(
        max_depth=None,
        max_features ='log2' , 
        min_samples_split=2,
        min_samples_leaf=2,
        random_state=42
    ))
])

dt_model.fit(X_train, y_train)
dt_preds = dt_model.predict(X_val)

dt_results = evaluate_model(y_val, dt_preds, model_name="Decision Tree")
```

### Decision Tree model

In this step, I train a Decision Tree regression model to predict store sales.

I use a `Pipeline` to combine the preprocessing step and the model in one structure. This is useful because the categorical variables are first transformed using the `preprocessor`, and then the transformed data is passed directly into the Decision Tree model.

The Decision Tree model learns patterns by splitting the data into smaller groups based on the feature values. For example, it can learn different sales patterns depending on the store, promotion status, day of the week, holidays, or store type.

The model is created with a few important settings:

- `max_depth=None` allows the tree to grow without a fixed depth limit.
- `max_features='log2'` means the model only considers a subset of features at each split, which can help reduce overfitting.
- `min_samples_split=2` means a node can be split if it has at least two samples.
- `min_samples_leaf=2` means each final leaf must contain at least two samples.
- `random_state=42` makes the results reproducible.

After defining the pipeline, I fit the model using the training data. Then I use it to predict sales on the validation set.

Finally, I evaluate the predictions using the same evaluation function as before, so the Decision Tree can be compared fairly with the baseline and other models.

## 11. XGBoost Model

```python
import os
import joblib
xgb_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', XGBRegressor(
        n_estimators=500,
        learning_rate=0.1,
        max_depth=10,
        subsample=0.8,
        colsample_bytree=0.6,
        min_child_weight=3,
        gamma=0.5,
        reg_alpha=0.1,
        reg_lambda=5,
        random_state=42,
        n_jobs=-1
    ))
])

# Train the model
xgb_model.fit(X_train, y_train)

# Predict on validation data
xgb_preds = xgb_model.predict(X_val)

# Evaluate the model
xgb_results = evaluate_model(y_val, xgb_preds, model_name="XGB")

# Save the trained XGBoost pipeline for the Streamlit app
os.makedirs("../models", exist_ok=True)
joblib.dump(xgb_model, "../models/rossmann_xgb_pipeline.pkl")

print("Model saved successfully to models/rossmann_xgb_pipeline.pkl")
```

### XGBoost model

In this step, I train an XGBoost regression model to predict store sales.

I use a `Pipeline` again so that preprocessing and modelling are handled together. The categorical features are first transformed by the `preprocessor`, and then the processed data is passed into the XGBoost model.

XGBoost is a gradient boosting algorithm. Instead of building one single model, it builds many decision trees one after another. Each new tree tries to correct the mistakes made by the previous trees. This usually makes XGBoost much stronger than a single Decision Tree.

The model is configured with several important parameters:

- `n_estimators=500` means the model builds 500 trees.
- `learning_rate=0.1` controls how much each tree contributes to the final prediction.
- `max_depth=10` allows each tree to learn fairly detailed patterns.
- `subsample=0.8` uses 80% of the training rows for each tree, which can help reduce overfitting.
- `colsample_bytree=0.6` uses 60% of the features for each tree.
- `min_child_weight=3` makes the model more conservative when creating new splits.
- `gamma=0.5` requires a split to provide enough improvement before it is accepted.
- `reg_alpha=0.1` and `reg_lambda=5` add regularization to make the model more stable and reduce overfitting.
- `random_state=42` makes the results reproducible.
- `n_jobs=-1` allows the model to use all available CPU cores during training.

After defining the pipeline, I train the model using the training data, then use it to predict sales on the validation set.

Finally, I evaluate the predictions using the same evaluation function as before. This allows me to compare XGBoost fairly with the baseline model and the Decision Tree model.

## 12. Random Forest Model

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', RandomForestRegressor(
        n_estimators=50,
        random_state=42,
        min_samples_split=10,
        min_samples_leaf=1,
        max_features='sqrt',
        max_depth=None,
        n_jobs=-1
    ))
])

rf_model.fit(X_train, y_train)

rf_preds = rf_model.predict(X_val)
rf_results = evaluate_model(y_val, rf_preds, model_name="Random Forest")
```

### Random Forest model

In this step, I train a Random Forest regression model to predict store sales.

I use a `Pipeline` so that preprocessing and modelling are connected in one workflow. The categorical columns are first converted into numerical format using the `preprocessor`, and then the processed data is passed into the Random Forest model.

Random Forest is an ensemble model, which means it builds many Decision Trees instead of only one. Each tree learns from a slightly different version of the training data, and the final prediction is based on the average prediction of all trees.

This usually makes Random Forest more stable than a single Decision Tree because it reduces the risk of relying too much on one tree’s mistakes.

The model is configured with these main settings:

- `n_estimators=50` means the forest builds 50 decision trees.
- `min_samples_split=10` means a node needs at least 10 samples before it can be split.
- `min_samples_leaf=1` means a final leaf can contain at least one sample.
- `max_features='sqrt'` means each tree only considers a subset of features at each split, which helps make the trees more diverse.
- `max_depth=None` allows the trees to grow without a fixed depth limit.
- `random_state=42` makes the results reproducible.
- `n_jobs=-1` allows the model to use all available CPU cores during training.

After defining the pipeline, I train the model on the training data. Then I use it to predict sales on the validation set.

Finally, I evaluate the predictions using the same evaluation function as before, so the Random Forest results can be compared fairly with the baseline, Decision Tree, and XGBoost models.

## 15. Feature Importance

```python
from sklearn.inspection import permutation_importance

perm_result = permutation_importance(
    xgb_model,
    X_val,
    y_val,
    n_repeats=1,
    random_state=42,
    scoring='neg_mean_absolute_error',
    n_jobs=1
)

feature_names = X_val.columns
perm_importance_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': perm_result.importances_mean
}).sort_values('Importance', ascending=False)

display(perm_importance_df.head(15))

plt.figure(figsize=(10, 6))
plt.barh(perm_importance_df['Feature'].head(15)[::-1], perm_importance_df['Importance'].head(15)[::-1])
plt.xlabel('Permutation Importance')
plt.ylabel('Feature')
plt.title('Top 15 Permutation Importances')
plt.show()
```

### Feature importance

This step checks which features have the strongest effect on the XGBoost model predictions.

Permutation importance works by randomly changing one feature at a time and measuring how much the model performance gets worse. If changing a feature causes a large drop in performance, that feature is considered important.

The results show that `Open` is the most important feature, which makes sense because sales are strongly affected by whether the store is open or closed.

The next most important features are mainly historical mean sales features, such as `MeanSales_ByStore_DayOfWeek_Promo`, `MeanSales_ByStore_Promo`, and `MeanSales_ByStore_DayOfWeek`. This shows that the model relies heavily on past sales patterns for each store, especially when combined with weekday and promotion information.

Other useful features include `DayOfWeek`, `Day`, `CompetitionDistance`, `Promo`, `StateHoliday`, `Store`, and `StoreType`.

Overall, the feature importance results show that the model is learning from both historical store behavior and business-related factors such as promotions, opening status, holidays, and competition distance.
